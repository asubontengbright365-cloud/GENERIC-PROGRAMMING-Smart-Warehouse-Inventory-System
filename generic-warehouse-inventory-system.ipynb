```python
from typing import Generic, TypeVar, List

T = TypeVar('T')

class Inventory(Generic[T]):      #Generic class to store inventory items of any type T. Supports storing multiple items while keeping type consistency.
    def __init__(self, category_name: str):
        self.category_name = category_name
        self._items: List[T] = [] # list to hold multiple items of type T

    def store(self, item: T) -> None:          #Store an item of type T
        self._items.append(item)

    def retrieve(self) -> List[T]:
        """Retrieve all stored items"""
        return self._items

    def display(self) -> None:            #Display the item type and all values
        print(f"Category: {self.category_name}")
        if self._items:
            data_type = type(self._items[0]).__name__
            print(f"Data Type: {data_type}")
        else:
            print("Data Type: Empty")
        print(f"Items: {self._items}")
        print(f"Total Items: {len(self._items)}")
        print("=" * 40)


# ---------------- DEMONSTRATION ---------------- #
print(" WAREHOUSE INVENTORY SYSTEM \n")

# 1. Inventory[str] for product names
product_inventory = Inventory[str]("Product Names")
product_inventory.store("HP Laptop")
product_inventory.store("Samsung Phone")
product_inventory.store("Textbook - Python")
product_inventory.store("Rice 5kg")
product_inventory.display()

# 2. Inventory[int] for stock quantities
stock_inventory = Inventory[int]("Stock Quantities")
stock_inventory.store(25)
stock_inventory.store(120)
stock_inventory.store(8)
stock_inventory.store(300)
stock_inventory.display()

# 3. Inventory[float] for product prices
price_inventory = Inventory[float]("Product Prices (GHS)")
price_inventory.store(3950.00)
price_inventory.store(150.50)
price_inventory.store(45.00)
price_inventory.store(12.99)
price_inventory.display()

# 4. Inventory[list] for batches of serial numbers
serial_inventory = Inventory[list]("Serial Number Batches")
serial_inventory.store(["SN-LAP-001", "SN-LAP-002", "SN-LAP-003"])
serial_inventory.store(["SN-PHN-101", "SN-PHN-102"])
serial_inventory.store(["SN-BOOK-55"])
serial_inventory.display()
```
